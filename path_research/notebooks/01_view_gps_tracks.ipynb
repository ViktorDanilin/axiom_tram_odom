# GPS-треки из ROS 2 bag

На карте показаны выбранные записи из `data/`. Источник координат — `/sensing/gnss/master/fix` (`sensor_msgs/msg/NavSatFix`). Для отображения используются ipyleaflet и подложка OpenStreetMap; при открытии карты нужен интернет.


```python
from pathlib import Path

import sys

NOTEBOOKS_DIR = Path.cwd() / "notebooks" if (Path.cwd() / "notebooks").is_dir() else Path.cwd()
sys.path.insert(0, str(NOTEBOOKS_DIR))
from tools.gps_tracks import load_gps_track
from tools.track_map import show_tracks_map

# Можно заменить ID записей на любые другие каталоги из data/.
BAGS = [
    "30618_0e41eac3",
    "30618_5eb8d2c9",
    "30618_437e855c",
]
MAX_POINTS = 1_200  # Точки для отрисовки; первая и последняя сохраняются.

# Работает и при запуске из корня проекта, и из каталога notebooks/.
DATA_DIR = next(
    (
        candidate
        for candidate in (Path.cwd() / "data_gps_union", Path.cwd().parent / "data_gps_union")
        if candidate.is_dir()
    ),
    None,
)
if DATA_DIR is None:
    raise FileNotFoundError(
        "Не найден каталог data_gps_union/. Запустите ноутбук из корня проекта или notebooks/."
    )

```

```python
tracks = [load_gps_track(DATA_DIR, bag_id, max_points=MAX_POINTS) for bag_id in BAGS]

for track in tracks:
    print(f"{track['id']}: {track['point_count']} GPS-точек, {track['length_m'] / 1000:.2f} км")
```

```python
gps_map = show_tracks_map(tracks)
```
